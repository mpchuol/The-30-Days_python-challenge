```python
# Get user input using input(“Enter your age: ”). If user is 18 or older, give feedback: You are old enough to drive. If below 18 give feedback to wait for the missing amount of years. Output:
age = int(input("Enter your age")) # it prompt the user to enter age
if age >= 18:
    print("You are old enough to drive.")
else:
    years_left = 18 - age # it calculate the missing years for one to drive
    print(f"You need {years_left} more years to learn to drive")
    
```

```python
my_age = int(input("Enter my age: "))  # Prompt the user to enter their age and convert it to an integer
your_age = int(input("Enter your age: "))  # Prompt the user to enter another person's age and convert it to an integer

if my_age > your_age: # Check if my_age is greater than your_age
    diff = my_age - your_age # Calculate the age difference
    if diff == 1:  # Check if the difference is exactly 1
        print(f"You are {diff} year older than me.")
    else:
        print(f"You are {diff} years older than me.")
elif my_age < your_age:  # Check if my_age is less than your_age
    diff = your_age - my_age # Calculate the age difference
    if diff == 1: # Check if the difference is exactly 1
        print(f"I am {diff} year older than you.")
    else:
        print(f"I am {diff} years older than you.")
else:  # If neither condition is met, both ages must be equal
    print("We are the same age!")
```

```python
# Get two numbers from the user using input prompt. If a is greater than b return a is greater than b, if a is less b return a is smaller than b, else a is equal to b. Output:
a = int(input("Enter the first number"))
b = int(input("Enter second number"))

if a > b:  # Check if 'a' is greater than 'b'
    print("a is greater than b") # Print the message if 'a' is greater
elif a < b:  # Check if 'a' is less than 'b'
    print("a is less than b")  # Print the message if 'a' is smaller
else: # If neither condition is met, 'a' and 'b' must be equal
    print("a is equal to b") # Print the message indicating equality
```

```python
 # Write a code which gives grade to students according to theirs scores:
student_score = int(input("Enter student score: ")) # Takes user input for the student's score and converts it to an integer.

def assign_grade(student_score): # function to determine the student's grade based on their score.
    if 80 <= student_score <= 100: # Checks if the score is between 80 and 100 to assign grade 'A'.
        return "A"
    elif 70 <= student_score <= 79: # Checks if the score is between 70 and 79 to assign grade 'B'.
        return "B"
    elif 60 <= student_score <= 69: # Checks if the score is between 60 and 69 to assign grade 'C'.
        return "C"
    elif 50 <= student_score <= 59: # Checks if the score is between 50 and 59 to assign grade 'D'.
        return "D"
    elif 0 <= student_score <= 49: # Checks if the score is between 0 and 49 to assign grade 'F'.
        return "F"
    else: # Handle cases where the score is outside the valid range.
        return "Invalid score"

print("Grade:", assign_grade(student_score)) # Prints the assigned grade based on the student's score.
```

```python
# Check if the season is Autumn, Winter, Spring or Summer. If the user input is: September, October or November, the season is Autumn. December, January or February, the season is Winter. March, April or May, the season is Spring June, July or August, the season is Summer
month = input("Enter a month: ")

def check_season(month): # efine a function that determines the season based on the given month.
    month = month.lower() # Convert the input month to lowercase to allow case-insensitive matching.
    if month in ["september", "october", "november"]: # Check if the month belongs to Autumn.
        return "Autumn"
    elif month in ["december", "january", "february"]: # Check if the month belongs to Winter.
        return "Winter"
    elif month in ["march", "april", "may"]: # Check if the month belongs to Spring.
        return "Spring"
    elif month in ["june", "july", "august"]:
        return "Summer"
    else: # handles invalid months
        return "Invalid month"

print("Season:", check_season(month)) # print the determined season.
```

```python
# If a fruit doesn't exist in the list add the fruit to the list and print the modified list. If the fruit exists print('That fruit already exist in the list')
fruits = ['banana', 'orange', 'mango', 'lemon'] # define a list

# Get user input and normalize it (strip spaces and convert to lowercase)
new_fruit = input("Enter a fruit: ").strip().lower()

# Check if the fruit is already in the list
if new_fruit in fruits:
    print("That fruit already exists in the list")  # Inform the user
else:
    fruits.append(new_fruit)  # Add the new fruit to the list
    print("Modified list:", fruits)  # Print the updated list
```
