```python
import numpy as np

# Define the function y = x^2 + 6x + 9
def quadratic_function(x):
    return x**2 + 6*x + 9

# Generate x values from -10 to 10
x_values = np.arange(-10, 11, 1)
y_values = [quadratic_function(x) for x in x_values]

# Solve for y = 0 (finding the root of the equation)
roots = np.roots([1, 6, 9])

# Print results
print("Computed values of y for different x values:")
for x, y in zip(x_values, y_values):
    print(f"y({x}) = {y}")

print("\nRoots where y = 0:", roots)
```
