Python_Type_Errors

```python
# Syntax Error
print "hello world" # missing parantheses in call to print is called Syntax Error 
```

```python
# Name Error
print(name) # when printed, it will display a Name Error because the name is not defined before being printed
```

```python
# Index Error
numbers = [2,3,4,6] # define a list
numbers[5] # to get element in index 5
print(numbers) # when printed, it will display an Index Error because the list has only indexes from 0 to 3, therefore it was out of range.append
```

```python
# Module Not Found Error
import maths # when printed, it will display a Module Not Found Error because maths is not found in module but math is found
```

```python
# Attribute Error
import math
math.PI # when printed, it display an attribute error because module math has no attribute "PI"
```

```python
# Key Error
details = {"name":"Maker", "age":55, "country":"Jamaica"} # define a dictionary
details["county"] # when printed, it will display a key error because of the typo by writing county instead of country
```

```python
# Type Error
x = 14 + "25" # int and str
print(x) # when printed, it will display a type error beacuse string and number cannot be add
```

```python
# Import Error
from math import power
power(3,6) # when printed, it will display an import error because import name "power" is not from "math" but pow it's found in "math"
```

```python
# Value Error
x = int("45b") # define a string
print(x) # when printed, it display a value error because we cannot change the string to a number because of "b" that's in it
```

```python
# Zero Division Error
x = 5 # int
y = 0 # int
result = x/y # it divide x by y
print(result) # when printed, it will display a zero division erro because a number can't be divide by 0
```
