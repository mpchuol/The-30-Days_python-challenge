```python
import numpy as np  # Importing NumPy for numerical operations
import pandas as pd  # Importing Pandas for data manipulation
import matplotlib.pyplot as plt  # Importing Matplotlib for data visualization

# 1. NumPy: Creating and Manipulating Arrays
arr = np.array([1, 2, 3, 4, 5])  # Creating a NumPy array
print("NumPy Array:", arr)  # Printing the array
print("Mean:", np.mean(arr))  # Calculating and printing the mean of the array
print("Standard Deviation:", np.std(arr))  # Calculating and printing the standard deviation

# 2. Pandas: Creating a DataFrame and Performing Operations
data = {
    'Name': ['Deng', 'Mary', 'Panom', 'David'],  # List of names
    'Age': [25, 30, 35, 40],  # Corresponding ages
    'Score': [85, 90, 88, 95]  # Corresponding scores
}
df = pd.DataFrame(data)  # Creating a Pandas DataFrame from the dictionary
print("\nPandas DataFrame:\n", df)  # Printing the DataFrame
print("\nAverage Score:", df['Score'].mean())  # Calculating and printing the average score

# 3. Matplotlib: Plotting Data
plt.figure(figsize=(6,4))  # Setting the figure size for the plot
plt.bar(df['Name'], df['Score'], color='blue')  # Creating a bar chart with names on the x-axis and scores on the y-axis
plt.xlabel('Name')  # Labeling the x-axis
plt.ylabel('Score')  # Labeling the y-axis
plt.title('Student Scores')  # Setting the title of the plot
plt.show()  # Displaying the plot
```
