```python
from flask import Flask, request, render_template_string
import threading
import re
from collections import Counter

app = Flask(__name__)

# Function to analyze text
def analyze_text(text):
    words = re.findall(r'\b\w+\b', text.lower())  
    num_words = len(words)
    num_chars = len(text)
    most_common = Counter(words).most_common(5)
    unique_words = len(set(words))
    lexical_density = (unique_words / num_words) * 100 if num_words > 0 else 0

    return {
        "num_words": num_words,
        "num_chars": num_chars,
        "most_common": most_common,
        "unique_words": unique_words,
        "lexical_density": round(lexical_density, 2)
    }

# Flask Routes
@app.route("/", methods=["GET", "POST"])
def home():
    text = ""
    results = None
    if request.method == "POST":
        text = request.form["text"]
        results = analyze_text(text)

    return render_template_string('''
    <!DOCTYPE html>
    <html lang="en">
    <head>
        <meta charset="UTF-8">
        <meta name="viewport" content="width=device-width, initial-scale=1.0">
        <title>Text Analyzer</title>
        <style>
            body { font-family: Arial, sans-serif; text-align: center; padding: 20px; }
            textarea { width: 80%; height: 150px; margin-bottom: 10px; }
            .container { max-width: 600px; margin: auto; }
        </style>
    </head>
    <body>
        <h2>Text Analyzer</h2>
        <form method="POST">
            <textarea name="text" placeholder="Enter your text here...">{{ text }}</textarea><br>
            <button type="submit">Analyze</button>
        </form>

        {% if results %}
        <div class="container">
            <h3>Results</h3>
            <p><strong>Word Count:</strong> {{ results.num_words }}</p>
            <p><strong>Character Count:</strong> {{ results.num_chars }}</p>
            <p><strong>Unique Words:</strong> {{ results.unique_words }}</p>
            <p><strong>Lexical Density:</strong> {{ results.lexical_density }}%</p>
            <h4>Most Common Words:</h4>
            <ul>
                {% for word, count in results.most_common %}
                    <li>{{ word }}: {{ count }}</li>
                {% endfor %}
            </ul>
        </div>
        {% endif %}
    </body>
    </html>
    ''', text=text, results=results)

# Function to run Flask inside Jupyter
def run_flask():
    app.run(host="127.0.0.1", port=5000, debug=True, use_reloader=False)

# Start Flask in a separate thread
threading.Thread(target=run_flask).start()
```
