```python
# Get the current day, month, year, hour, minute and timestamp from datetime module
from datetime import datetime # import the datetime module
now = datetime.now() # it get the current date and time
# Extract and store individually components
day = now.day # it get the current day of the month
month = now.month # get the current month
year = now.year # it get the current year
hour = now.hour # it get the current hour(24-hr format)
minute = now.minute # it get the current minute
timestamp = now.timestamp() # it get the current timestamp

print(f"Day: {day}") # Display the current day
print(f"Month: {month}") # Display the current month
print(f"Year: {year}")  # Display the current year
print(f"Hour: {hour}") # Display the current hour
print(f"Minute: {minute}") # Display the current minute
print(f"Timestamp: {timestamp}") # Display the times
```

```python
# Format the current date using this format: "%m/%d/%Y, %H:%M:%S")
from datetime import datetime  
now = datetime.now() # Get the current date and time
formatted_date = now.strftime("%m/%d/%Y, %H:%M:%S") # Format the date and time
print(f"Formatted Date: {formatted_date}") # it print the formatted result
```

```python
# Today is 5 December, 2019. Change this time string to time.
from datetime import datetime
date_string = "5 December, 2019" # define a date string
date_object = datetime.strptime(date_string, "%d %B, %Y") # it convert string to datetime object
print(date_object) # it print the result
```

```python
# Calculate the time difference between now and new year.
from datetime import datetime
current_time = datetime.now() # it get the current time
new_year = datetime(current_time.year + 1, 1, 1) # set the upcoming new year's date
time_diff = new_year - current_time # it calculate the time difference
print(f"Time remaining untill New Year:{time_diff}") # it print the time difference time untill next year
```

```python
# Calculate the time difference between 1 January 1970 and now.
from datetime import datetime
initial_time = datetime(1970, 1, 1) # define the year, 1 January 1970
current_time = datetime.now() # it get the current time
time_diff = current_time - initial_time # it calculate the time difference
print(f"The Time difference since 1 January 1970:{time_diff}") # it print the time difference
```

```python
# Sheduling and Reminder
# it used to set reminders and countdown
from datetime import datetime
event_date = datetime(2025, 8, 7) # define the date for event
current_time = datetime.now() # it get the current date
time_remaining = event_date - current_time # it get the remaining time for event to happen
print(f"Time remaining days Until event day:{time_remaining}") # it print the remaining time for event to happen
```
