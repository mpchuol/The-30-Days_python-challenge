```python
# Create a function that finds the ten most spoken languages

import json                          # Import the json module to work with JSON files
from collections import Counter     # Import Counter from collections to count occurrences of items

# Define a function that finds the ten most spoken languages from a JSON file
def ten_most_spoken_languages(json_file_path):
    try:
        # Open the JSON file at the specified path in read mode with UTF-8 encoding
        with open(json_file_path, 'r', encoding='utf-8') as file:
            # Load the content of the file into a Python object (usually a list of dictionaries)
            countries_data = json.load(file)
        
        # Create an empty list to store all languages spoken in the dataset
        all_languages = []

        # Loop through each country's data in the list
        for country in countries_data:
            # Add the country's languages to the all_languages list
            # 'get' is used to avoid KeyError if 'languages' key is missing
            all_languages.extend(country.get('languages', []))

        # Count how many times each language appears (i.e., in how many countries it's spoken)
        language_counts = Counter(all_languages)

        # Get the 10 most common languages and their counts as a list of tuples
        most_common_languages = language_counts.most_common(10)

        # Display the results
        print("Top 10 most spoken languages (by number of countries using them):")
        for language, count in most_common_languages:
            print(f"{language}: {count} countries")

        # Return the result in case we want to use it elsewhere
        return most_common_languages

    # Handle the case where the file path is incorrect or file does not exist
    except FileNotFoundError:
        print(f"File not found: {json_file_path}")

    # Handle cases where the file is not in valid JSON format
    except json.JSONDecodeError:
        print("Invalid JSON format.")

    # Handle any other unexpected errors
    except Exception as e:
        print(f"An error occurred: {e}")

# Call the function with the path to your countries_data.json file
ten_most_spoken_languages(r"C:\Users\pc\Desktop\30 DAYS OF PYTHON CHALLENGE\Day_19\Countries_data.json")
```

```python
# create a function that creates a list of the ten most populated countries

import json # Import the json module to read JSON data
from operator import itemgetter # Import itemgetter to help sort dictionaries by a specific key

# Define a function that finds the ten most populated countries
def ten_most_populated_countries(json_file_path):
    try:
        # Open the JSON file using the given file path, with UTF-8 encoding
        with open(json_file_path, 'r', encoding='utf-8') as file:
            # Load the JSON content into a Python list of dictionaries
            countries_data = json.load(file)

        # Create a new list of dictionaries with only 'country' name and 'population'
        # Use get() to avoid errors if keys are missing; provide default values just in case
        population_list = [
            {"country": country.get("name", "Unknown"), "population": country.get("population", 0)}
            for country in countries_data
        ]

        # Sort the list of countries by the 'population' key in descending order
        # Then slice the list to keep only the top 10 countries
        top_10 = sorted(population_list, key=itemgetter("population"), reverse=True)[:10]

        # Print the result in a user-friendly format
        print("Top 10 most populated countries:")
        for item in top_10:
            # Print each country with its population, formatted with commas
            print(f"{item['country']}: {item['population']:,} people")

        # Return the top 10 list in case it’s needed for further processing
        return top_10

    # Handle the error if the file doesn't exist at the specified path
    except FileNotFoundError:
        print(f"File not found: {json_file_path}")
    
    # Handle the error if the JSON content is not formatted correctly
    except json.JSONDecodeError:
        print("Invalid JSON format.")
    
    # Catch and print any other types of unexpected errors
    except Exception as e:
        print(f"An error occurred: {e}")

# Call the function with the actual path to your JSON file
ten_most_populated_countries(r"C:\Users\pc\Desktop\30 DAYS OF PYTHON CHALLENGE\Day_19\Countries_data.json")
```

```python
import re
import os  # For path validation

def extract_incoming_emails(file_path):
    """
    Extracts all unique email addresses from a file.
    Args:
        file_path (str): Path to the input file.
    Returns:
        list: Unique email addresses found, or None if errors occur.
    """
    try:
        # 1. Validate file existence and readability
        if not os.path.exists(file_path):
            raise FileNotFoundError(f"File not found: {file_path}")
        if not os.access(file_path, os.R_OK):
            raise PermissionError(f"No read permissions for: {file_path}")

        # 2. Read file content with explicit encoding (UTF-8)
        with open(file_path, 'r', encoding='utf-8') as f:
            content = f.read()
            print(f"File read successfully. Size: {len(content)} characters.")  # Debugging

        # 3. Regex pattern with stricter boundary checks
        email_pattern = r'(?<!\S)[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}(?!\S)'
        
        # 4. Find all matches and validate
        email_list = re.findall(email_pattern, content)
        if not email_list:
            print("Warning: No emails found. Check file content or pattern.")
            return []

        # 5. Deduplicate and sort for consistency
        unique_emails = sorted(list(set(email_list)))  # Sort for readability

        # 6. Output results
        print(f"\nTotal unique emails found: {len(unique_emails)}")
        print("Sample emails (first 5):")
        for email in unique_emails[:5]:
            print(f"  - {email}")

        return unique_emails

    except Exception as e:
        print(f"\nError: {str(e)}")
        return None

# ===== TEST THE FUNCTION =====
if __name__ == "__main__":
    # Test with your file path (update this!)
    test_path = (r"C:\Users\pc\Desktop\30 DAYS OF PYTHON CHALLENGE\Day_19\email_exchange.txt")
    
    # Verify path before running
    print(f"\nAttempting to read: {test_path}")
    if os.path.exists(test_path):
        print("Path exists. Proceeding...\n")
        emails = extract_incoming_emails(test_path)
    else:
        print("Error: Invalid file path. Please update `test_path`.")
```

```python
import os
import re
from collections import Counter
from math import sqrt

# Function to clean text: remove punctuation, convert to lowercase
def clean_text(text):
    # Convert to lowercase
    text = text.lower()
    # Remove punctuation and numbers
    text = re.sub(r'[^a-z\s]', '', text)
    return text

# Function to remove stopwords from the text
def remove_stopwords(text, stopwords):
    # Tokenize the text into words
    words = text.split()
    # Filter out the stopwords
    filtered_words = [word for word in words if word not in stopwords]
    return filtered_words

# Function to calculate cosine similarity between two texts
def check_text_similarity(text1, text2):
    # Count word frequencies for both texts
    counter1 = Counter(text1)
    counter2 = Counter(text2)

    # Get the set of all unique words in both texts
    all_words = set(counter1.keys()).union(set(counter2.keys()))

    # Build vectors
    vec1 = [counter1.get(word, 0) for word in all_words]
    vec2 = [counter2.get(word, 0) for word in all_words]

    # Compute cosine similarity
    dot_product = sum(a * b for a, b in zip(vec1, vec2))
    magnitude1 = sqrt(sum(a * a for a in vec1))
    magnitude2 = sqrt(sum(b * b for b in vec2))

    if magnitude1 == 0 or magnitude2 == 0:
        return 0.0  # Avoid division by zero

    return dot_product / (magnitude1 * magnitude2)

# ================================
# MAIN LOGIC TO COMPARE TWO FILES
# ================================

try:
    # Load stopwords from file (one word per line)
    with open('data/stopwords.txt', 'r', encoding='utf-8') as sw_file:
        stopwords = set(sw_file.read().splitlines())

    # Load the first speech
    with open(r"C:\Users\pc\Desktop\30 DAYS OF PYTHON CHALLENGE\Day_19\Michele.txt", 'r', encoding='utf-8') as f1:
        michelle_text = f1.read()

    # Load the second speech
    with open(r"C:\Users\pc\Desktop\30 DAYS OF PYTHON CHALLENGE\Day_19\Melina.txt", 'r', encoding='utf-8') as f2:
        melina_text = f2.read()

    # Clean the texts (pass content, not file paths)
    michelle_clean = clean_text(michelle_text)
    melina_clean = clean_text(melina_text)

    # Remove stopwords
    michelle_words = remove_stopwords(michelle_clean, stopwords)
    melina_words = remove_stopwords(melina_clean, stopwords)

    # Check similarity
    similarity = check_text_similarity(michelle_words, melina_words)

    # Display result as a percentage
    print(f"Similarity between Michelle's and Melina's speech: {similarity * 100:.2f}%")

except FileNotFoundError as e:
    print(f"Error: File not found - {e}")
except Exception as e:
    print(f"An error occurred: {e}")
```
