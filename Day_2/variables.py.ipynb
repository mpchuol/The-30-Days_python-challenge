```python
# Declaring different variables
first_name = "Panom" # string
last_name = "Chuol" # string
full_name = first_name + " " + last_name # string concatenation
country = "Bukina Faso" # string
city = "Kigali" # string
age = 35 # integer
year = 2024 # integer
is_married = True # boolean
is_true = True # boolean
is_light_on = False # boolean

# declaring multiple variables in one line
person_name,person_age,person_country = "Deng", 45,"South Africa" # string, integer, string

print(f"First Name: {first_name}") # it print the value of first_name variable
print(f"Last Name: {last_name}") # it print the value of last_name variable
print(f"Full Name: {full_name}") # it print the value of full_name variable
print(f"Country: {country}") # it print the value of country 
print(f"City: {city}") # it print the value of city
print(f"Age: {age}") # it print the value of age
print(f"Year: {year}") # it print the value of year
print(f"Is Married: {is_married}") # it print the value of is_married
print(f"Is True: {is_true}") # it print the value of is_true
print(f"Is Light On: {is_light_on}") # it print the value of is_light_on
print(f"His name is {person_name} and he is {person_age} year-old. He was born in {person_country}") # it print the value of person_name, person_age and person_country

```

```python
# Checking the data type of the variables
first_name = "Panom" # string
last_name = "Chuol" # string
full_name = first_name + " " + last_name # string concatenation
country = "Bukina Faso" # string
city = "Kigali" # string
age = 35 # integer
year = 2024 # integer
is_married = True # boolean
is_true = True # boolean
is_light_on = False # boolean

# declaring multiple variables in one line
person_name,person_age,person_country = "Deng", 45,"South Africa" # string, integer, string

print(type(first_name)) # it print the data type
print(type(last_name)) # it print the data type
print(type(full_name)) # it print the data type
print(type(country)) # it print the data type
print(type(city)) # it print the data type
print(type(age)) # it print the data type
print(type(year)) # it print the data type
print(type(is_married)) # it print the data type
print(type(is_true)) # it print the data type
print(type(is_light_on)) # it print the data type
print(type(person_name)) # it print the data type
print(type(person_age)) # it print the data type
print(type(person_country)) # it print the data type
```

```python
# Uisng the len function
first_name = "Panom" # string
print(len(first_name)) # it print the length of the string
```

```python
# Comparing the length of two strings
first_name = "Panom" # string
last_name = "Chuol" # string
if len(first_name) > len(last_name): # if the length of first name is greater than the last name
    print("First name is longer than last name") # it print the message
elif len(first_name) == len(last_name): # if the length of first and last are equal
    print("First name and last name are equal") # it print the message
else: # if the length of first name is less than the last name
    print("last name is longer than first name") # it print the message

    
```

```python
# Adding two numbers
num_one = 5 # integer
num_two = 4 # integer
total = num_one + num_two # it add the two numbers
print(total) # it print the total
```

```python
# subtracting two numbers
num_one = 5 # integer
num_two = 4 # integer
diff = num_one - num_two # it subtract the two numbers
print(diff) # it print the difference
```

```python
# multiplying two numbers
num_one = 5 # integer
num_two = 4 # integer
product = num_one * num_two # it multiply the two numbers
print(product) # it print the product
```

```python
# dividing two numbers
num_one = 5 # integer
num_two = 4 # integer
division = num_one / num_two # it divide the two numbers
print(division) # it print the division
```

```python
# Use modulus division
num_one = 5 # integer
num_two = 4 # integer
remainder = num_two % num_one # it find the remainder
print(remainder) # it print the remainder
```

```python
# Power of numbers
num_one = 5 # integer
num_two = 4 # integer
exp = num_one ** num_two # it find the power of the two numbers
print(exp) # it print the power
```

```python
# Find floor division 
num_one = 5 # integer
num_two = 4 # integer
floor_division = num_one // num_two # it find the floor division
print(floor_division) # it print the floor division
```

```python
# The radius of a circle is 30 meters
radius = 30 # integer
area_of_circle = 3.142 * radius ** 2 # it find the area of the circle
print(area_of_circle) # it print the area of the circle
```

```python
# Circumference of a circle
radius = 30 # integer
circum_of_circle = 2 * 3.142 * radius # it find the circumference of the circle
print(circum_of_circle) # it print the circumference of the circle
```
