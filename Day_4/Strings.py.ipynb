```python
# Create an acronym or an abbreviation for the name 'Python For Everyone'
def generate_acronym(phrase):
    words = phrase.split() # it split the phrase into words
    acronym = "".join(word[0].upper() for word in words) # it get first letter of each word 
    return acronym
phrase = "Python For Everyone" # define the phrase
acronym = generate_acronym(phrase) # it call the function
print(acronym) # it print PFE as the output
 
```

```python
# Use the string formatting method to display the following:
#radius = 10
#area = 3.14 * radius ** 2
#The area of a circle with radius 10 is 314 meters square.
radius = 10 # define radius of the circle
area = 3.14 * radius ** 2 # how to calculate the area of a circle
print("The area of a circle with radius {} is {:.0f} meters square.".format(radius, area)) #Print the formatted result using an f-string. # {:.0f} ensures that the area is displayed as a whole number (no decimals)
```

```python
# Make the following using string formatting methods:
a = 8 # Define number
b = 6 # Define number

# Perform calculations and format output
print("{} + {} = {}".format(a, b, a + b))
print("{} - {} = {}".format(a, b, a - b))
print("{} * {} = {}".format(a, b, a * b))
print("{} / {} = {:.2f}".format(a, b, a / b))  # Format to 2 decimal places
print("{} % {} = {}".format(a, b, a % b))
print("{} // {} = {}".format(a, b, a // b))
print("{} ** {} = {}".format(a, b, a ** b))
```
