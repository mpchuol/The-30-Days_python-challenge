```python
# Filter only negative and zero in the list using list comprehension
numbers = [-4, -3, -2, -1, 0, 2, 4, 6]

filtered_numbers = [num for num in numbers if num <= 0]  # Keep only negative numbers and zero

print(filtered_numbers) # it print only the numbers that are <=0 
```

```python
# Flatten the following list of lists of lists to a one dimensional list 
list_of_lists = [[[1, 2, 3]], [[4, 5, 6]], [[7, 8, 9]]]

flattened_list = [num for sublist in list_of_lists for inner_list in sublist for num in inner_list]

print(flattened_list)  # it print all the values of list in one list
```

```python
# Using list comprehension create the following list of tuples
table = [(i, 1, i, i**2, i**3, i**4, i**5) for i in range(11)] # # Generate a list of tuples using list comprehension

for row in table: # # Loop through the list and print each tuple
    print(row) # # Print each generated tuple
```

```python
# Flatten the following list to a new list:countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]
# Given list of countries with their capitals
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]

# Flatten and transform the list using list comprehension
flattened_countries = [[country.upper(), country[:3].upper(), capital.upper()] for sublist in countries for country, capital in sublist]

print(flattened_countries) # Print the transformed list
```

```python
 # Given list of countries with their capitals
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]

# Convert to a list of dictionaries
countries_dict_list = [{'country': country, 'capital': capital} for sublist in countries for country, capital in sublist]

# Print each dictionary separately
for country in countries_dict_list:
    print(country)  # Print each dictionary on a new line

# Print the entire list of dictionaries 
print("\nComplete list:", countries_dict_list)
```

```python
# Change the following list of lists to a list of concatenated strings:
# Given list of names
names = [[('Asabeneh', 'Yetayeh')], [('David', 'Smith')], [('Donald', 'Trump')]]

# Convert to a list of concatenated strings
concatenated_names = [f"{first} {last}" for sublist in names for first, last in sublist]

# Print the result
print(concatenated_names)
```

```python
# Write a lambda function which can solve a slope or y-intercept of linear functions.
# Lambda function to calculate the slope (m) of a linear equation
slope = lambda x1, y1, x2, y2: "Undefined slope (vertical line)" if x1 == x2 else (y2 - y1) / (x2 - x1)

# Lambda function to calculate the y-intercept (b) using y = mx + b => b = y - mx
y_intercept = lambda x1, y1, x2, y2: "No y-intercept (vertical line)" if x1 == x2 else y1 - slope(x1, y1, x2, y2) * x1

x1, y1, x2, y2 = 2, 3, 5, 9
print("Slope:", slope(x1, y1, x2, y2))  # Output: 2.0
print("Y-intercept:", y_intercept(x1, y1, x2, y2))  # Output: -1.0
print("Slope:", slope(4, 2, 4, 8))  # Output: "Undefined slope (vertical line)"
print("Y-intercept:", y_intercept(4, 2, 4, 8))  # Output: "No y-intercept (vertical line)"
```
