```python
# Find the median age (one middle item or two middle items divided by two)
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24] # Define the list of ages

# Sort the list
ages.sort()
print(f"Sorted list:{ages}")

# Find the median
n = len(ages)  # Get the length of the list
if n % 2 == 1:
    median = ages[n // 2]  # Odd number of elements
else:
    median = (ages[n // 2 - 1] + ages[n // 2]) / 2  # Even number of elements

# Print the result
print(f"Median age: {median}")
```

```python
# Find the average age (sum of all items divided by their number )
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24] # Define the list of ages

# Calculate the average
average_age = sum(ages) / len(ages) # total of values is divided by their length

# Print the result
print(f"Average age: {average_age}") # it print the average
```

```python
# Find the range of the ages (max minus min)
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]

maximum_ages = max(ages) # it get the maximum age in the list
minimum_ages = min(ages) # it get the minimum age in the list
print(f"Maximum age:{maximum_ages}") # it print the maximum age
print(f"Minimum age:{minimum_ages}") # it print the minimum age

# Calculate the average
average_age = sum(ages) / len(ages) # total of values is divided by their length

# Print the result
print(f"Average age: {average_age}") # it print the average
# Find the range
age_range = maximum_ages - minimum_ages # it get the range between the two

# Print the result
print(f"Range of ages: {age_range}") # it print the difference between the two
```

```python
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]

maximum_ages = max(ages) # it get the maximum age in the list
minimum_ages = min(ages) # it get the minimum age in the list
print(f"Maximum age:{maximum_ages}") # it print the maximum age
print(f"Minimum age:{minimum_ages}") # it print the minimum age

# Find the range
age_range = max(ages) - min(ages) # it get the range between the two

# Print the result
print(f"Range of ages: {age_range}") # it print the difference between the two


# Compare differences
if minimum_ages > maximum_ages:
    print("Min is farther from the average than Max.")
elif minimum_ages < maximum_ages:
    print("Max is farther from the average than Min.")
else:
    print("Both distances are equal.")
```
