```python
# Get the value of skills and check the data type, it should be a list
student_dict = {
    "first_name":"Panom",
    "last_name":"Dend",
    "gender":"male",
    "age":20,
    "marital status":"single",
    "skills":["technichian","Java"],
    "country":"Jamaica",
    "city":"Kigali"
} # define a dictionary
skills_values = student_dict["skills"] # it extract only the skills' values
print(skills_values) # it print the values for skills
print(type(skills_values)) # it print the data type of the skills values
```

```python
# Get the dictionary keys as a list
student_dict = {
    "first_name":"Panom",
    "last_name":"Dend",
    "gender":"male",
    "age":20,
    "marital status":"single",
    "skills":["technichian","Java"],
    "country":"Jamaica",
    "city":"Kigali"
} # define a dictionary
key_list = list(student_dict.keys()) # it convert the dictionary to a list and get the keys of the list
print(key_list) # it print only the keys
```

```python
# Get the dictionary values as a list
student_dict = {
    "first_name":"Panom",
    "last_name":"Dend",
    "gender":"male",
    "age":20,
    "marital status":"single",
    "skills":["technichian","Java"],
    "country":"Jamaica",
    "city":"Kigali"
} # define a dictionary
values_list = list(student_dict.values()) # it convert the dictionary to a list and get the values of the list
print(values_list) # it print only the values
```

```python
# Change the dictionary to a list of tuples using items() method
student_dict = {
    "first_name":"Panom",
    "last_name":"Dend",
    "gender":"male",
    "age":20,
    "marital status":"single",
    "skills":["technichian","Java"],
    "country":"Jamaica",
    "city":"Kigali"
} # define a dictionary
student_tuples = list(student_dict.items()) # it convert dictionary to a list of tuples
print(student_tuples) # it print the output
```

```python
# Delete one of the items in the dictionary
student_dict = {
    "first_name":"Panom",
    "last_name":"Dend",
    "gender":"male",
    "age":20,
    "marital status":"single",
    "skills":["technichian","Java"],
    "country":"Jamaica",
    "city":"Kigali"
} # define a dictionary
student_dict.pop("skills") # it delete the skills and its values
print(student_dict) # it print the keys without skills
```
