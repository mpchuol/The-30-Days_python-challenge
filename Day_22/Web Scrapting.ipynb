```python
import requests  # Import requests to send HTTP requests
from bs4 import BeautifulSoup  # Import BeautifulSoup to parse HTML content
import json  # Import json to store data in JSON format

def fetch_page(url):
    """Fetches and returns the webpage content."""
    try:
        response = requests.get(url, timeout=10)  # Send an HTTP GET request with a timeout
        response.raise_for_status()  # Raise an error for HTTP request failures
        return response.content
    except requests.RequestException as e:
        print(f"Error fetching the webpage: {e}")  # Print the error and exit
        return None

def extract_data(soup):
    """Extracts required statistics from the parsed HTML."""
    data = {}  # Initialize an empty dictionary to store extracted data

    # Example: Extracting multiple statistics safely
    sections = {
        'Research Expenditures': 'h3',
        'Student Enrollment': 'h3',
        'Faculty Members': 'h3',
        'Campuses': 'h3'
    }

    for key, tag in sections.items():
        section = soup.find(tag, text=key)
        if section:
            value = section.find_next_sibling('p')
            data[key] = value.text.strip() if value else 'Not Available'
        else:
            data[key] = 'Not Available'  # Handle missing sections

    return data

def save_to_json(data, filename='bu_facts_stats.json'):
    """Saves extracted data to a JSON file."""
    with open(filename, 'w') as json_file:
        json.dump(data, json_file, indent=4)
    print(f"\n✅ Data successfully saved to '{filename}'")

# Main Execution
url = 'http://www.bu.edu/president/boston-university-facts-stats/'  # Define the target URL
page_content = fetch_page(url)

if page_content:
    soup = BeautifulSoup(page_content, 'html.parser')  # Parse the webpage content using BeautifulSoup
    data = extract_data(soup)  # Extract data

    # Print extracted data before saving
    print("\n📊 Extracted Boston University Facts & Stats:\n")
    for key, value in data.items():
        print(f"{key}: {value}")

    save_to_json(data)  # Save data to JSON file
```

```python
import requests  # Import requests to send HTTP requests
from bs4 import BeautifulSoup  # Import BeautifulSoup to parse HTML content
import json  # Import json to store extracted data in JSON format

# Define the new target URL
url = "https://archive.ics.uci.edu/datasets"

try:
    response = requests.get(url, timeout=10)  # Send an HTTP GET request with a 10-second timeout
    response.raise_for_status()  # Raise an HTTPError for bad responses (4xx, 5xx)
except requests.RequestException as e:
    print(f"Error fetching the webpage: {e}")  # Print error message if request fails
    exit()

# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(response.content, "html.parser")

datasets = []  # Initialize an empty list to store dataset information

# Locate dataset containers (Adjust this selector based on actual page structure)
dataset_elements = soup.find_all("div", class_="dataset-card")

if not dataset_elements:
    print("Error: No datasets found on the page. The page structure might have changed.")
    exit()

# Extract dataset information
for dataset in dataset_elements:
    name = dataset.find("h3").text.strip() if dataset.find("h3") else "Unknown"
    description = dataset.find("p").text.strip() if dataset.find("p") else "No description available"
    link = dataset.find("a")["href"] if dataset.find("a") else "No link available"
    
    datasets.append({
        "Name": name,
        "Description": description,
        "URL": f"https://archive.ics.uci.edu{link}"  # Construct full URL
    })

# Save data to JSON file
json_filename = "uci_datasets.json"
try:
    with open(json_filename, "w", encoding="utf-8") as json_file:  # Open file in write mode
        json.dump(datasets, json_file, indent=4, ensure_ascii=False)  # Save JSON with indentation
    print(f"✅ Data successfully extracted and saved to '{json_filename}'")
except Exception as e:
    print(f"Error writing to file: {e}")
```

```python
import requests
from bs4 import BeautifulSoup
import json

# Wikipedia URL for the U.S. Presidents list
url = "https://en.wikipedia.org/wiki/List_of_presidents_of_the_United_States"

try:
    response = requests.get(url, timeout=10)
    response.raise_for_status()
except requests.RequestException as e:
    print(f"Error fetching the webpage: {e}")
    exit()

soup = BeautifulSoup(response.content, "html.parser")

# Locate the main table containing U.S. Presidents
table = soup.find("table", class_="wikitable")

if not table:
    print("Error: Could not find the presidents' table. The page structure might have changed.")
    exit()

presidents = []

# Extract table rows
rows = table.find_all("tr")[1:]  # Skip the header row

for row in rows:
    cols = row.find_all(["td", "th"])
    
    if len(cols) < 5:  # Ensure row has enough columns
        continue
    
    # Extract relevant details
    number = cols[0].text.strip()  # President number
    name = cols[1].text.strip().split("\n")[0]  # President name (handle links and extra text)
    term = cols[2].text.strip()  # Term duration
    party = cols[3].text.strip()  # Political party
    vice_president = cols[4].text.strip()  # Vice President(s)

    presidents.append({
        "Number": number,
        "Name": name,
        "Term": term,
        "Party": party,
        "Vice President(s)": vice_president
    })

# Save data to JSON file
json_filename = "us_presidents.json"
with open(json_filename, "w", encoding="utf-8") as json_file:
    json.dump(presidents, json_file, indent=4)

print(f"✅ Data successfully extracted and saved to '{json_filename}'")
```
