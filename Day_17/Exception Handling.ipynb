```python
names = ["Finland","Sweden","Norway","Denmark","Iceland","Estonia","Russia"] # define a list with countries
*nordic_countries, es, ru = names # unpacking the countries
print(f"Nordic countries:{nordic_countries}") # it print the first five countries after being unpacked
print(f"Second last country:{es}") # it print the second last country in the list after being unpacked
print(f"Last country:{ru}") # it print the last country in the list after being unpacked
```
