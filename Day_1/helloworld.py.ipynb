```python
# Finding Euclidean distance between two points
import math # importing math module
x1, y1 = 2, 3 # first point
x2, y2 = 10, 8 # second point
# Euclidean distance formula
distance = math.sqrt((x2-x1)**2 + (y2-y1)**2) # it calculate the distance between the two points
print(f"The Euclidean distance = {distance}") # it print the output of the calculated distance
```
