```python
# Explain the difference between map, filter, and reduce.
# map() it transofrm each element
# filter() selects elements from an iterable that satisfy a given condition
# reduce() it aggregates the list to a single value 
```

```python
# Higher-Order Function is a function that takes another function as an argument or returns a function as a result.
# Closure is a function that remembers the variables from its enclosing scope even after the outer function has finished executing.
# Decorator is a special type of higher-order function that modifies or extends the behavior of another function without changing its structure.
```

```python
# Define a call function before map, filter or reduce, see examples.
# Using map()
def square(num): # function to square a number
    return num ** 2

numbers = [1, 2, 3, 4, 5] # define a list
squared_numbers = list(map(square, numbers)) # Applying the square function to each element using map()

print(squared_numbers)  # it print the squared numbers
```

```python
# Using filter()
def is_even(num): # Function to check if a number is even
    return num % 2 == 0
numbers = [1, 2, 3, 4, 5, 6] # define a list
even_numbers = list(filter(is_even, numbers)) # Filtering even numbers using filter()
print(even_numbers) # it print only even numbers
```

```python
# Using reduce()
from functools import reduce
def add(x, y): # Function to add two numbers
    return x + y
numbers = [1, 2, 3, 4, 5] # define a list
sum_result = reduce(add, numbers) # Applying reduce to sum the numbers
print(sum_result)  # it print the sum of numbers
```

```python
# Chain two or more list iterators (eg. arr.map(callback).filter(callback).reduce(callback))
from functools import reduce

# Given list of words
words = ["hello", "John", "life", "is", "a", "precious", "gift"]

# Chaining map, filter, and reduce
sentence = reduce(
    lambda acc, word: acc + " " + word,  # Concatenate words
    filter(
        lambda word: len(word) > 2,  # Keep words longer than 2 characters
        map(lambda word: word.capitalize(), words)  # Capitalize each word
    )
)

# Print the final sentence
print(sentence)
```

```python
# Use reduce to concatenate all the countries and to produce this sentence: Estonia, Finland, Sweden, Denmark, Norway, and Iceland are north European countries
from functools import reduce  
countries = ["Estonia", "Finland", "Sweden", "Denmark", "Norway", "Iceland"] # define a list

# Using reduce to concatenate country names into a sentence
sentence = reduce(lambda x, y: x + ", " + y, countries[:-1]) + f", and {countries[-1]} are north European countries."
print(sentence) # it print the result
```

```python
# 
def count_countries_by_letter(countries):
     
    country_count = {}
    
    for country in countries:
        first_letter = country[0].upper()  # Get the first letter and convert to uppercase
        country_count[first_letter] = country_count.get(first_letter, 0) + 1  # Count occurrences
    return country_count
countries = ["Finland", "France", "Sweden", "Denmark", "Norway", "Netherlands", "Nigeria", "India", "Iceland", "Italy"]
country_dict = count_countries_by_letter(countries)
print(country_dict) # it print the result
```

```python
from collections import Counter

# Data: List of countries with population and languages spoken
countries = [
    {"name": "China", "population": 1444216107, "languages": ["Mandarin"]},
    {"name": "India", "population": 1393409038, "languages": ["Hindi", "English"]},
    {"name": "United States", "population": 331893745, "languages": ["English"]},
    {"name": "Indonesia", "population": 273523621, "languages": ["Indonesian"]},
    {"name": "Pakistan", "population": 220892331, "languages": ["Urdu", "English"]},
    {"name": "Brazil", "population": 212559409, "languages": ["Portuguese"]},
    {"name": "Nigeria", "population": 206139587, "languages": ["English", "Hausa", "Yoruba", "Igbo"]},
    {"name": "Bangladesh", "population": 164689383, "languages": ["Bengali"]},
    {"name": "Russia", "population": 143932377, "languages": ["Russian"]},
    {"name": "Mexico", "population": 128932753, "languages": ["Spanish"]},
    {"name": "Japan", "population": 125836021, "languages": ["Japanese"]},
    {"name": "Ethiopia", "population": 114963588, "languages": ["Amharic"]},
    {"name": "Philippines", "population": 109581085, "languages": ["Filipino", "English"]},
    {"name": "Egypt", "population": 102334403, "languages": ["Arabic"]},
    {"name": "Vietnam", "population": 97338583, "languages": ["Vietnamese"]},
    {"name": "Turkey", "population": 84339067, "languages": ["Turkish"]},
    {"name": "Iran", "population": 83992953, "languages": ["Persian"]},
    {"name": "Germany", "population": 83783945, "languages": ["German"]},
    {"name": "Thailand", "population": 69799978, "languages": ["Thai"]},
    {"name": "France", "population": 65273512, "languages": ["French"]},
]

# Step 1: Extract all languages spoken
all_languages = []
for country in countries:
    all_languages.extend(country.get('languages', []))  # Get languages safely

# Step 2: Count occurrences of each language
language_counts = Counter(all_languages)

# Step 3: Get the 10 most spoken languages
most_spoken_languages = language_counts.most_common(10)

# Step 4: Get the 10 most populated countries
most_populated_countries = sorted(countries, key=lambda x: x['population'], reverse=True)[:10]

# Display results
print("\nTop 10 Most Spoken Languages by Location:")
for lang, count in most_spoken_languages:
    print(f"{lang}: {count} countries")

print("\nTop 10 Most Populated Countries:")
for country in most_populated_countries:
    print(f"{country['name']}: {country['population']} people")
```
