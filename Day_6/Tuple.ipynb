```python
# Slice out the middle item or items from the food_stuff_tp tuple or food_stuff_lt list.
fruits = ("mango", "banana", "apple", "watermelon") # define a tuple
vegetables = ("spinach", "cabbage", "sukumawiki") # define a tuple
animal_products = ("mutton", "beef", "honey") # define a tuple

# Join all tuples into one
food_stuff_tp = fruits + vegetables + animal_products
print("Full Tuple:", food_stuff_tp)

# Find the length of the tuple
length = len(food_stuff_tp)

# Slicing out the middle item(s)
if length % 2 == 1:  # Odd length → One middle item
    middle_item = food_stuff_tp[length // 2]
else:  # Even length → Two middle items
    middle_item = food_stuff_tp[length // 2 - 1: length // 2 + 1]

print("Middle item(s):", middle_item) # it print the middle items
```
