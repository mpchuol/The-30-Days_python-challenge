```python
# Add 'Twitter' to it_companies
it_companies = {"Facebook","Google","Microsoft","Apple","IBM","Oracle","Amazon"} # define a set
it_companies.add("Twitter") # it add the twitter the set
print(it_companies) # it print the set
```

```python
# Insert multiple IT companies at once to the set it_companies
it_companies = {"Facebook","Google","Microsoft","Apple","IBM","Oracle","Amazon"} # define a set
it_companies.update({"Messenger","Whatsapp","Macbook","Safaricom"}) # it add all these multiples items at once
print(it_companies) # it print the updated set
```

```python
# Remove one of the companies from the set it_companies
it_companies = {"Facebook","Google","Microsoft","Apple","IBM","Oracle","Amazon"} # define a set
it_companies.remove("Facebook") # it remove the facebook from the set
print(it_companies) # it print the set without facebook
```

```python
# What is the difference between remove and discard
it_companies = {"Facebook","Google","Microsoft","Apple","IBM","Oracle","Amazon"} # define a set
it_companies.remove("facebook") # it remove the facebook from the set
print(it_companies) # it will raised a keyerror because the element facebook is not found

it_companies = {"Facebook","Google","Microsoft","Apple","IBM","Oracle","Amazon"} # define a set
it_companies.discard("Facebook") # it remove the facebook from the set
print(it_companies) # it will not raised a keyerror when the element is not found
```

```python
# Join A and B
A = {19, 22, 24, 20, 25, 26} # define a set
B = {19, 22, 20, 25, 26, 24, 28, 27} # define a set
C = A.union(B) # it merged the sets while removing duplicates since sets only sotre unique values
print(C) # it print the set without the duplicates
```

```python
# Find A intersection B
A = {19, 22, 24, 20, 25, 26} # define a set
B = {19, 22, 20, 25, 26, 24, 28, 27} # define a set
C = A.intersection(B) # it returns a set of items which are in both the set A and B
print(C) # it print the items that present in both sets
```

```python
# Is A subset of B
A = {19, 22, 24, 20, 25, 26} # define a set
B = {19, 22, 20, 25, 26, 24, 28, 27} # define a set
is_subset = A.issubset(B) # it check if A is a subset of B
print(is_subset) # it print True because all the items in A are in B
```

```python
# Are A and B disjoint sets
A = {19, 22, 24, 20, 25, 26} # define a set
B = {19, 22, 20, 25, 26, 24, 28, 27} # define a set
C = A.isdisjoint(B) # it check if the sets have common items, if so, then it will be a join set
print(C) # it will print false because there are common items in the two sets
```

```python
# Join A with B and B with A
A = {19, 22, 24, 20, 25, 26} # define a set
B = {19, 22, 20, 25, 26, 24, 28, 27} # define a set
# A union B
A_union_B = A.union(B) # it merge the two sets without duplicates
print(A_union_B) # it print the set without duplicates

# B union A
B_union_A = B.union(A) # it merge the two sets without
print(B_union_A) # it print the se without duplicates
# The output is the same
```

```python
# What is the symmetric difference between A and B
A = {19, 22, 24, 20, 25, 26} # define a set
B = {19, 22, 20, 25, 26, 24, 28, 27} # define a set
sym_diff = A.symmetric_difference(B) # it check the values of symmentric difference the two sets have
print(sym_diff) # it print only the items that are not found in set A but found in set B
```

```python
# Convert the ages to a set and compare the length of the list and the set, which one is bigger?
age = [22, 19, 24, 25, 26, 24, 25, 24]
my_set = set(age) # it convert the list to a set

# comparing the lengths 
print(f"The length of list:{len(age)}") # it print the length of the list
print(f"The length of set: {len(my_set)}") # it print the length of the set

# Which one is bigger?
if len(age) > len(my_set): # it check if the len of list is > len of a set
    print("The list is bigger.")
else:
    print("The set is bigger.")
```

```python
# I am a teacher and I love to inspire and teach people. How many unique words have been used in the sentence? Use the split methods and set to get the unique words.
sentence = "I am a teacher and I love to inspire and teach people." # define a string
words = sentence.split() # it split the sentence into words
unique_words = set(words) # it convert the string to a set
print(f"The unique words are:{unique_words}") # it print the unique words
print(f"The length of the unique words:{len(unique_words)}") # it print the length of the unique words
```
