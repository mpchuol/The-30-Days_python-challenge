```python
import numpy as np  # Import the NumPy library for numerical operations

# Generate 100 random numbers from a normal distribution with mean=5 and std=0.5
np_normal_dis = np.random.normal(5, 0.5, 100)

# Calculate and print the minimum value in the array
print('min: ', np_normal_dis.min())  

# Calculate and print the maximum value in the array
print('max: ', np_normal_dis.max())  

# Calculate and print the mean (average) value of the array
print('mean: ', np_normal_dis.mean())  

# Calculate and print the median value of the array
print('median: ', np.median(np_normal_dis))  # Use np.median() for median calculation

# Calculate and print the standard deviation of the array
print('sd: ', np_normal_dis.std())
```

```python
import numpy as np  # Import NumPy library

# Creating a 2D NumPy array (example with random values)
two_dimension_array = np.random.randint(1, 100, size=(5, 4))  
# A 5x4 array with random integers between 1 and 100

print(two_dimension_array)  # Print the entire 2D array

# Find and print the minimum values for each column (axis=0 means column-wise)
print('Column with minimum: ', np.amin(two_dimension_array, axis=0))  

# Find and print the maximum values for each column
print('Column with maximum: ', np.amax(two_dimension_array, axis=0))  

print('=== Row ===')  # Separator for clarity

# Find and print the minimum values for each row (axis=1 means row-wise)
print('Row with minimum: ', np.amin(two_dimension_array, axis=1))  

# Find and print the maximum values for each row
print('Row with maximum: ', np.amax(two_dimension_array, axis=1))
```

```python
import numpy as np  # Import NumPy library

a = [1, 2, 3]  # Define a simple list

# Repeat the entire list two times
print('Tile:   ', np.tile(a, 2))  
# Output: [1 2 3 1 2 3]  -> The whole list is repeated twice

# Repeat each element in the list two times
print('Repeat: ', np.repeat(a, 2))  
# Output: [1 1 2 2 3 3]  -> Each element is repeated twice before moving to the next
```

```python
import numpy as np  # Import NumPy library

# Define two 1D NumPy arrays
f = np.array([1, 2, 3])
g = np.array([4, 5, 3])

# Compute the dot product
# Calculation: (1*4) + (2*5) + (3*3) = 4 + 10 + 9 = 23
result = np.dot(f, g)

print(result)  # Output: 23
```

```python
import numpy as np  # Import NumPy library

# Define two 2x2 matrices
h = [[1, 2], [3, 4]]
i = [[5, 6], [7, 8]]

# Perform matrix multiplication
result = np.matmul(h, i)

print(result)
```

```python
import numpy as np  # Import NumPy library

# Define a 2x2 matrix
i = np.array([[5, 6], [7, 8]])

# Compute the determinant of the matrix
# Formula for a 2x2 matrix:
# det(A) = (ad - bc) for a matrix [[a, b], [c, d]]
# det(i) = (5*8 - 7*6) = 40 - 42 = -2
determinant = np.linalg.det(i)

print(determinant)  # Output: -2.0
```

```python
import numpy as np  # Import NumPy for numerical operations
import matplotlib.pyplot as plt  # Import Matplotlib for plotting

# Define a NumPy array with temperature values
temp = np.array([1, 2, 3, 4, 5])

# Calculate pressure using the formula: pressure = 2 * temp + 5
pressure = temp * 2 + 5

# Create a line plot of temperature vs pressure
plt.plot(temp, pressure, marker='o', linestyle='-')  # Adding markers for better visibility

# Label the x-axis and y-axis
plt.xlabel('Temperature in °C')  # Use '°' symbol instead of 'oC' for correct formatting
plt.ylabel('Pressure in atm')

# Add a title to the graph
plt.title('Temperature vs Pressure')

# Set x-axis ticks from 0 to 5.5 with a step of 0.5
plt.xticks(np.arange(0, 6, step=0.5))

# Display the plot
plt.show()
```

```python
import numpy as np  # Import NumPy for numerical operations
import seaborn as sns  # Import Seaborn for visualization
import matplotlib.pyplot as plt  # Import Matplotlib for plotting

# Define parameters for the normal distribution
mu = 28        # Mean
sigma = 15     # Standard deviation
samples = 100000  # Number of samples

# Generate random samples from a normal distribution
x = np.random.normal(mu, sigma, samples)

# Plot the distribution using Seaborn
ax = sns.histplot(x, kde=True, bins=50)  # Use sns.histplot() instead of sns.distplot()

# Set axis labels
ax.set(xlabel="x", ylabel="y")

# Display the plot
plt.show()
```
