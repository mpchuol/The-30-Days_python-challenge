```python
# What is the most frequent word in the following paragraph?
from collections import Counter
import re
paragraph = "I love teaching. If you do not love teaching what else can you love. I love Python if you do not love something which can give you all the capabilities to develop an application what else can you love." # define a string
words = re.findall(r'\b\w+\b', paragraph.lower()) # it remove punctuation and convert to lowercase
words_counts = Counter(words) # count the word occurrences
most_frequent = words_counts.most_common(1)[0] # it get the most frequent word
print(f"The most frequent word is {most_frequent[0]} with {most_frequent[1]} occurrences.") # it print the most frequent word in the paragraph and with how many occurrences

```

```python
# Extract these numbers from this whole text and find the distance between the two furthest particles.
import re

text = """The position of some particles on the horizontal x-axis are -12, -4, -3 and -1 in the negative direction, 
0 at origin, 4 and 8 in the positive direction.""" # define string

points = list(map(int, re.findall(r'-?\d+', text))) # Extract numbers from the text
sorted_points = sorted(points) # Sort the points in ascending order
distance = sorted_points[-1] - sorted_points[0] # Calculate distance between the two furthest points
print(f"Extracted points: {points}") # it printed the extracted numbers
print(f"Sorted points: {sorted_points}") # it printed the sorted numbers
print(f"Distance between the furthest particles: {distance}") # it print the distance
```

```python
# Write a pattern which identifies if a string is a valid python variable
import re
import keyword

def is_valid_variable(var_name):
    pattern = r'^[a-zA-Z_][a-zA-Z0-9_]*$'  # Regex pattern for valid variable names
    return bool(re.match(pattern, var_name)) and var_name not in keyword.kwlist

# Test cases
print(is_valid_variable('first_name'))  # True
print(is_valid_variable('first-name'))  # False
print(is_valid_variable('1first_name'))  # False
print(is_valid_variable('firstname'))  # True
print(is_valid_variable('def'))  # False (Python keyword)
print(is_valid_variable('_variable'))  # True
```

```python
#Clean the following text. After cleaning, count three most frequent words in the string.
import re
from collections import Counter

sentence = '''%I $am@% a %tea@cher%, &and& I lo%#ve %tea@ching%;. There $is nothing; &as& mo@re rewarding as educa@ting &and& @emp%o@wering peo@ple. ;I found tea@ching m%o@re interesting tha@n any other %jo@bs. %Do@es thi%s mo@tivate yo@u to be a tea@cher!?'''

# Step 1: Clean the text (remove special characters)
cleaned_text = re.sub(r'[^a-zA-Z\s]', '', sentence)

words = cleaned_text.lower().split() # Convert to lowercase and split into words

word_counts = Counter(words) # it count word occurrences

most_common_words = word_counts.most_common(3) # Get the three most common words

# Display results
print("Cleaned text:", cleaned_text)
print("\nTop 3 most frequent words:")
for word, count in most_common_words:
    print(f"{word}: {count} times")
```
