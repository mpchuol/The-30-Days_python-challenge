```python
import pandas as pd  # Import Pandas for data handling
import os  # Import os to check file existence

# Define the file path
file_path = r"C:\Users\pc\Desktop\30 DAYS OF PYTHON CHALLENGE\Day_25\hacker_new.csv"

# Check if the file exists before reading
if os.path.exists(file_path):
    # Read the CSV file from the 'data' directory
    df = pd.read_csv(file_path)

    # Display basic dataset info
    print("✅ Dataset Overview:")
    print(df.info())  # Show column names, non-null counts, data types

    # Count the number of rows and columns
    num_rows, num_columns = df.shape
    print(f"\n✅ Number of Rows: {num_rows}")
    print(f"✅ Number of Columns: {num_columns}")

    # Display summary statistics
    print("\n✅ Summary Statistics:")
    print(df.describe(include='all'))  # Shows numerical and categorical data summary

    # Display first five rows
    print("\n✅ First 5 Rows:")
    print(df.head())

    # Extract 'title' column if it exists
    if 'title' in df.columns:
        title_series = df['title']  # Extracts the 'title' column
        print("\n✅ First 5 Titles:")
        print(title_series.head())  # Display first 5 titles

        # Filter titles that contain 'Python' (case-insensitive)
        python_titles = df[df['title'].str.contains("Python", case=False, na=False)]
        print(f"\n✅ Titles Containing 'Python' ({len(python_titles)} results):")
        print(python_titles['title'].head())

        # Filter titles that contain 'JavaScript' (case-insensitive)
        js_titles = df[df['title'].str.contains("JavaScript", case=False, na=False)]
        print(f"\n✅ Titles Containing 'JavaScript' ({len(js_titles)} results):")
        print(js_titles['title'].head())

    else:
        print("❌ Error: 'title' column not found in the dataset.")

else:
    print(f"❌ Error: File '{file_path}' not found. Please check the path.")
```
